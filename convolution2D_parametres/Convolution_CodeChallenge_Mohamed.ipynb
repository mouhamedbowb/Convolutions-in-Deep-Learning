# Exemple résolu

### Convolution d'une image 1x256x256 pour produire un résultat 1x252x84

```python
# parametresinChans  = 1imsize   = [256,256]outChans = 1krnSize  = 7 # nombre impair de preferencestride   = (1,3)padding  = 1# creation de l'instancec = nn.Conv2d(inChans,outChans,krnSize,stride,padding)# creation d'une imageimg = torch.rand(1,inChans,imsize[0],imsize[1])# execution de la convolution et calcul de sa formeresimg = c(img)empSize = torch.squeeze(resimg).shape# calcul de la taille attendue selon la formuleexpectSize = np.array([outChans,0,0],dtype=int)expectSize[1] = np.floor( (imsize[0]+2*padding-krnSize)/stride[0] ) + 1expectSize[2] = np.floor( (imsize[1]+2*padding-krnSize)/stride[1] ) + 1# verificationprint(f'Taille attendue : {expectSize}')print(f'Taille obtenue  : {list(empSize)}')
```
